```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data=pd.read_csv('diabetes.csv')
data
```

```python
for j in data.columns:
    try:
        pd.to_numeric(data[j])
        data[j] = data[j].fillna(data[j].mean())
    except:
        pass
data.info()
data.head(5)
```

```python
corr_matrix=data[['Glucose','BloodPressure','Insulin','Age','BMI','Outcome']].corr()
corr_matrix
```

```python
sns.heatmap(corr_matrix,annot=True,cmap='rainbow')
plt.title('Correlation Matrix')
plt.show()
```

```python
max_corr_outcome=corr_matrix['Outcome'].drop('Outcome')
print(f'Max Correlation with Outcome is {max(max_corr_outcome)}.')
print(f'Feature with the most correlation is {max_corr_outcome.idxmax()}.')
```
