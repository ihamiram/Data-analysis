```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
tempreture=np.random.randint(15,40,180)
humidity=np.random.randint(20,100,180)
rainfall=np.random.randint(0,50,180)
Days=np.tile(np.arange(1,31,1),6)
cities=['tehran','shiraz','isfahan','tabriz','mashhad','ahvaz']*30
cities=np.sort(cities)
df = pd.DataFrame(
    {'tempreture': tempreture, 'Humidity': humidity, 'Rainfall': rainfall, 'Cities': cities, 'Days': Days})
df
```

```python
avg_rht_per_city=df.groupby('Cities')[['tempreture','Humidity','Rainfall']].mean()
#rain humidity tempreture = rht
avg_rht_per_city
```

```python
df_avg_temp=df.groupby('Cities')['tempreture'].mean().sort_values()
print(df_avg_temp.index[0],df_avg_temp.index[-1])
```

```python
df_count_rainfall=df[df['Rainfall']>10]
df_count_rainfall=df_count_rainfall['Rainfall'].count()
df_count_rainfall
```

```python
df_of_isfahan=df[df['Cities']=='isfahan']
plt.scatter(df_of_isfahan['tempreture'],df_of_isfahan['Humidity'])
plt.title('Isfahan Tempreture vs Humidity')
plt.xlabel('Tempreture (°C)')
plt.ylabel('Humidity(%)')
plt.show()
```
